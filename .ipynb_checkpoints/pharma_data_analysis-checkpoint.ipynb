1)What are the total sales quantities for each drug category (ATC code)?

```python
df.columns
```

```python
print('    |The total sales for each Drug category|   ')
sales_of_all=df[['M01AB','M01AE','N02BA','N02BE','N05B','N05C','R03','R06']].sum()
sales_of_all
```

2)Which individual drug brands have the highest total sales?

```python
drug_higest_sales=sales_of_all[sales_of_all==sales_of_all.max()]
print(' |The Drug with the highest Sales| \n ',drug_higest_sales)
```

```python
top_sales_values=sales_of_all.sort_values(ascending=False)
top_sales_values
```

```python
sns.barplot(x=top_sales_values.index,y=top_sales_values.values,palette='rainbow')
plt.title('Highest Drug Sales')
plt.xlabel('Drug Categories')
plt.ylabel('Total Sales')
```

3)Which three drugs have the highest sales in January 2015, July 2016, September 2017.

```python
sales_month_wise=df.groupby(['Month','Year'])[['M01AB', 'M01AE', 'N02BA', 'N02BE', 'N05B', 'N05C', 'R03','R06']].sum()
sales_month_wise
```

```python
jan_2015_a= sales_month_wise.loc[(1,2015)]
jan_2015_b=jan_2015_a.sort_values(ascending=False).head(3)
jan_2015_b
```

```python
jul_2016_a= sales_month_wise.loc[(7,2016)]
jul_2016_b=jul_2016_a.sort_values(ascending=False).head(3)
jul_2016_b
```

```python
sept_2017_a= sales_month_wise.loc[(9,2017)]
sept_2017_b=sept_2017_a.sort_values(ascending=False).head(3)
sept_2017_b
```

```python
plt.figure(figsize=(16,4))
plt.subplot(1,3,1)

sns.barplot(x=jan_2015_b.index,y=jan_2015_b.values,palette='rainbow')
plt.title('Top 3 Drugs January 2015')
plt.xlabel('Drug Categories')
plt.ylabel('Total Sales')

plt.subplot(1,3,2)
sns.barplot(x=jul_2016_b.index,y=jul_2016_b.values,palette='rainbow')
plt.title('Top 3 Drugs July 2016')
plt.xlabel('Drug Categories')
plt.ylabel('Total Sales')

plt.subplot(1,3,3)
sns.barplot(x=sept_2017_b.index,y=sept_2017_b.values,palette='rainbow')
plt.title('Top 3 Drugs September 2017')
plt.xlabel('Drug Categories')
plt.ylabel('Total Sales')
```

4)Which drug has sold the most often in 2017?

```python
sales_year_wise=df.groupby(['Year'])[['M01AB', 'M01AE', 'N02BA', 'N02BE', 'N05B', 'N05C', 'R03','R06']].sum()
sales_year_wise
```

```python
sales_2017=sales_year_wise.loc[2017].sort_values(ascending=False)
print('The Drug Sold Most Often in 2017 is :',sales_2017.head(1).index[0])
```

5)Which drug category has the highest average daily sales?

```python
sales_year_wise_avg=df[['M01AB', 'M01AE', 'N02BA', 'N02BE', 'N05B', 'N05C', 'R03','R06']].mean()
print('The Drug with the highest average daily sales :',sales_year_wise_avg.sort_values(ascending=False).head(1).index[0])
```

```python
plt.figure(figsize=(16, 8))

years = sales_R03.index.get_level_values('Year').unique()

for i, year in enumerate(years, 1):
    sales_year = sales_R03.loc[year]

    plt.subplot(2, 3, i)

    sns.barplot(
        x=sales_year.index,
        y=sales_year.values,
        palette='rainbow'
    )

    plt.title(f'R03 Sales - {year}')
    plt.xlabel('Month')
    plt.ylabel('Total Sales')

plt.tight_layout()
plt.show()
```

From the following data,which maps the Sales of R03(Respiratory drugs) for all 12 months,across the years 2014-2019,we can infer that R03 has more sales during the months of october,november,december,january.This is a effect of severe air pollution spikes and changing winter weather months,which matches the data.
